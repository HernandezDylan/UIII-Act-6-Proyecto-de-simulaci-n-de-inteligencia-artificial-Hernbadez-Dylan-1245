#### Pago contactless

```python
print("==========================================")
print("SISTEMA INTELIGENTE DE PAGO CONTACTLESS")
print("==========================================")

while True:

    print("\n=========== MENU ===========")
    print("1 Realizar pago")
    print("2 Verificar seguridad")
    print("3 Consultar saldo")
    print("4 Analizar tarjeta")
    print("5 Salir")

    opcion = input("Selecciona opcion: ")

    # OPCION 1
    if opcion == "1":

        print("\n====== REALIZAR PAGO ======")

        nombre = input("Nombre del cliente: ")

        tarjeta = input("Tipo de tarjeta: ")

        saldo = float(input("Saldo disponible: "))

        pago = float(input("Monto del pago: "))

        print("\n====== RESULTADO ======")

        if saldo >= pago:

            restante = saldo - pago

            print("PAGO APROBADO")

            print("Cliente:", nombre)

            print("Tarjeta:", tarjeta)

            print("Saldo restante:", restante)

        elif pago > saldo:

            print("PAGO RECHAZADO")

            print("Saldo insuficiente")

        else:

            print("Error en el sistema")


    # OPCION 2
    elif opcion == "2":

        print("\n====== ANALISIS DE SEGURIDAD ======")

        seguridad = int(input("Nivel de seguridad del 1 al 10: "))

        intentos = int(input("Intentos incorrectos detectados: "))

        if seguridad >= 8 and intentos <= 2:

            print("SEGURIDAD ALTA")

            print("Sistema protegido")

        elif seguridad >= 5:

            print("SEGURIDAD MEDIA")

            print("Supervision necesaria")

        else:

            print("SEGURIDAD BAJA")

            print("Posible riesgo detectado")


    # OPCION 3
    elif opcion == "3":

        print("\n====== CONSULTAR SALDO ======")

        nombre = input("Nombre del cliente: ")

        saldo = float(input("Saldo actual: "))

        if saldo >= 5000:

            print("Saldo alto")

        elif saldo >= 1000:

            print("Saldo medio")

        else:

            print("Saldo bajo")

        print("Cliente:", nombre)

        print("Saldo:", saldo)


    # OPCION 4
    elif opcion == "4":

        print("\n====== ANALISIS DE TARJETA ======")

        tarjeta = input("Tipo de tarjeta: ")

        compras = int(input("Cantidad de compras del dia: "))

        if tarjeta == "Visa":

            print("Tarjeta Visa detectada")

        elif tarjeta == "MasterCard":

            print("Tarjeta MasterCard detectada")

        else:

            print("Otra tarjeta detectada")


        if compras >= 15:

            print("Actividad sospechosa")

        elif compras >= 5:

            print("Actividad normal")

        else:

            print("Pocas compras registradas")


    # OPCION 5
    elif opcion == "5":

        print("Programa finalizado")

        break


    # OPCION INCORRECTA
    else:

        print("Opcion incorrecta")
```
